```python
course = (
    "CS101",
    ["Ahammed", "Rahul", "Sneha", "Arjun"],
    [85, 78, 92, 88]
)

course_id, students, grades = course

# Display info
print("Course ID:", course_id)
print("Students:", students)
print("Grades:", grades)

# Average grade
avg = sum(grades) / len(grades)
print("Average grade:", avg)

# Highest grade student
max_grade = max(grades)
top_student = students[grades.index(max_grade)]
print("Top student:", top_student, "-", max_grade)

# Update grade (example: Rahul → 82)
target = "Rahul"
new_grade = 82

if target in students:
    idx = students.index(target)
    grades[idx] = new_grade

print("Updated grades:", grades)
```

```python
employees = [
    ("Ahammed", 20, 40),
    ("Rahul", 25, 35),
    ("Sneha", 30, 30)
]

# Total salary for each
salaries = [(name, rate * hours) for name, rate, hours in employees]

print("Salaries:", salaries)

# Highest salary
top_employee = max(salaries, key=lambda x: x[1])
print("Highest salary:", top_employee)

# Average salary
avg_salary = sum(s for _, s in salaries) / len(salaries)
print("Average salary:", avg_salary)
```

```python
students = {
    "Ahammed": (85, 90, 88),
    "Rahul": (78, 82, 80),
    "Sneha": (92, 95, 94)
}

# Average per student
averages = {name: sum(scores)/len(scores) for name, scores in students.items()}

print("Averages:", averages)

# Highest average
top_student = max(averages, key=averages.get)
print("Top student:", top_student, "-", averages[top_student])

# Average per subject
subject_avg = [
    sum(scores[i] for scores in students.values()) / len(students)
    for i in range(3)
]

print("Subject averages:", subject_avg)
```

```python
## to calculate CAGR 

def calculate_cagr(initial_value, final_value, years):
    if initial_value <= 0 or final_value <= 0 or years <= 0:
        raise ValueError("All inputs must be positive numbers.")
    
    cagr = (final_value / initial_value) ** (1 / years) - 1
    return cagr

result = calculate_cagr(1000, 2000, 5)
```

```python
initial = 1000
final = 2000
years = 5

cagr = (final / initial) ** (1 / years) - 1

print("CAGR is:", cagr * 100, "%")
```
