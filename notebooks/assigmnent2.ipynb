```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Generate random salaries
np.random.seed(42)
salaries = np.random.randint(50000, 100001, 1000)

# Create DataFrame
df = pd.DataFrame({
    'Salary': salaries
})

######### Descriptive Statistics #########

print("Descriptive Statistics\n")
print(df.describe())

print("\nMedian Salary:", df["Salary"].median())
print("Mode Salary:", df["Salary"].mode()[0])
print("Standard Deviation:", df["Salary"].std())
print("Skewness:", df["Salary"].skew())
```

```python
# Histogram
plt.figure(figsize=(8,5))
sns.histplot(df["Salary"], bins=20, kde=True)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
mean_salary = df["Salary"].mean()
median_salary = df["Salary"].median()
mode_salary = df["Salary"].mode()[0]

plt.figure(figsize=(8,5))

sns.histplot(df["Salary"], bins=20, kde=True, edgecolor="black")

# Mean
plt.axvline(mean_salary, color='red', linestyle='--', label=f"Mean: {mean_salary:.2f}")

# Median
plt.axvline(median_salary, color='green', linestyle='--', label=f"Median: {median_salary}")

# Mode
plt.axvline(mode_salary, color='blue', linestyle='--', label=f"Mode: {mode_salary}")

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.legend()

plt.show()
```

```python
# Independent Sample t-Test

import numpy as np
from scipy import stats

np.random.seed(42)

# Sample data
# Scores of students using two methods

method_A = np.random.normal(loc=75, scale=8, size=35)
method_B = np.random.normal(loc=78, scale=7, size=35)

# Independent two-sample t-test

t_stat, p_value = stats.ttest_ind(
    method_A,
    method_B,
    equal_var=False
)

print("Mean A:", np.mean(method_A))
print("Mean B:", np.mean(method_B))

alpha = 0.05

if p_value < alpha:
    print("Reject null hypothesis: Means are significantly different")
else:
    print("Fail to reject null hypothesis: No significant difference between means")
```

```python
import numpy as np
from scipy import stats

np.random.seed(42)

# Sample data
# Before and after training scores

before = np.random.normal(loc=75, scale=6, size=30)

after = before + np.random.normal(loc=4, scale=3, size=30)

# Paired t-test

t_stat, p_value = stats.ttest_rel(before, after)

print("Mean Before:", np.mean(before))
print("Mean After:", np.mean(after))

print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject null hypothesis: Scores changed significantly after training")
else:
    print("Fail to reject null hypothesis: No significant change after training")
```
