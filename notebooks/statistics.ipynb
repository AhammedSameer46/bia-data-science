```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
from scipy import stats
from collections import Counter

# Nice default settings for plots

plt.rcParams.update({
    "figure.figsize": (10, 5),
    "axes.titlesize": 14,
    "axes.labelsize": 17,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

sns.set_palette("Set2")
display("Libraries imported successfully!")
```

```python
flg, axes = plt.subplots(1,2,figsize=(13,5))

blood_groups = ["A","B","AB","O"]
counts = [30,25,10,35]
axes[0].bar(blood_groups, counts, color=sns.color_palette("Set2"))
axes[0].set_title("Categorical (Nominal) - Blood Group Disturbution")
axes[0].set_xlabel("Blood Group")
axes[0].set_ylabel("Count")
for i,v in enumerate(counts):
    axes[0].text(i,v + 0.5, str(v),ha="center", fontweight="bold")


np.random.seed(42)
heights = np.random.normal(loc=170, scale=10, size=300)
axes[1].hist(heights , bins=25, edgecolor="white", color="steelblue", alpha=0.85)
axes[1].set_title("Numberical (Continuous)- Height Distribution (cm)")
axes[1].set_xlabel("Height (cm)")
axes[1].set_ylabel("Frequency")


plt.suptitle("Types of Data", fontsize=16 ,  fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```
