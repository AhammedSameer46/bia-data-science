```python
df = df.apply(lambda col: col.str.strip() if col.dtype == "object" else col)
```

```python
df = df.drop(columns=["country", "date_added"])
```

```python
df.info()
```

```python
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Not Rated")
```

```python
df.isnull().sum()
```

```python
df["title"].duplicated().sum()
```

```python
df[df["title"].duplicated(keep=False)].sort_values("title")
```

```python
df[["duration_value", "duration_unit"]] = df["duration"].str.extract(r"(\d+)\s+(\w+)")

df["duration_value"] = df["duration_value"].astype(int)
```

```python
df[["duration", "duration_value", "duration_unit"]].head(10)
```

```python
df["type"].value_counts()
```

```python
df["rating"].value_counts()
```

```python
df["duration_unit"].value_counts()
```

```python
df["duration_unit"] = df["duration_unit"].replace({
    "Season": "Seasons"
})
```

```python
df["duration_unit"].value_counts()
```

```python
df["rating"] = df["rating"].replace({
    "UNRATED": "Not Rated",
    "NOT_RATE": "Not Rated",
    "ALL_AGES": "ALL",
    "AGES_16_": "16+",
    "AGES_18_": "18+",
    "16": "16+"
})
```

```python
df["rating"].value_counts()
```

```python
df["release_decade"] = (df["release_year"] // 10) * 10
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe(include="all")
```

```python
sorted(df["rating"].unique())
```

```python
df.drop(columns=["duration"], inplace=True)
```

```python
df["primary_genre"] = df["listed_in"].str.split(",").str[0].str.strip()
```

```python
def duration_category(x):
    if x < 60:
        return "Short"
    elif x <= 120:
        return "Medium"
    else:
        return "Long"

df["duration_category"] = df["duration_value"].apply(duration_category)
```

```python
df.info()
df.describe(include="all")
df.isnull().sum()
```

```python
# Check unique values
print(df["type"].unique())
print(df["duration_unit"].unique())
print(sorted(df["rating"].unique()))

# Check duplicates again
print(df.duplicated().sum())

# Check missing values
print(df.isnull().sum().sum())
```
